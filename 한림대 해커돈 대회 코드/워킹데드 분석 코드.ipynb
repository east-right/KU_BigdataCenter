```python
Hr_sig = Input(shape=(600,6))

Hr_sig_a = Conv1D(16, 6, activation='relu')(Hr_sig)
Hr_sig_a = BatchNormalization()(Hr_sig_a) 
Hr_sig_a = MaxPooling1D(3, 3)(Hr_sig_a) 

Hr_sig_a = Conv1D(32, 6, activation='relu')(Hr_sig_a)
Hr_sig_a = BatchNormalization()(Hr_sig_a) 
Hr_sig_a = MaxPooling1D(3, 3)(Hr_sig_a) 

Hr_sig_a = Conv1D(64, 6, activation='relu')(Hr_sig_a)
Hr_sig_a = BatchNormalization()(Hr_sig_a) 
Hr_sig_a = MaxPooling1D(3, 3)(Hr_sig_a)

Hr_sig_a = Conv1D(128, 6, activation='relu')(Hr_sig_a)
Hr_sig_a = BatchNormalization()(Hr_sig_a) 
Hr_sig_a = MaxPooling1D(3, 3)(Hr_sig_a)
```

```python
Hr_sig1 = Input(shape=(600,6))

Hr_sig_b = Conv1D(16, 6, activation='relu')(Hr_sig1)
Hr_sig_b = BatchNormalization()(Hr_sig_b) 
Hr_sig_b = MaxPooling1D(3, 3)(Hr_sig_b) 

Hr_sig_b = Conv1D(32, 6, activation='relu')(Hr_sig_b)
Hr_sig_b = BatchNormalization()(Hr_sig_b) 
Hr_sig_b = MaxPooling1D(3, 3)(Hr_sig_b) 

Hr_sig_b = Conv1D(64, 6, activation='relu')(Hr_sig_b)
Hr_sig_b = BatchNormalization()(Hr_sig_b) 
Hr_sig_b = MaxPooling1D(3, 3)(Hr_sig_b)

Hr_sig_b = Conv1D(128, 6, activation='relu')(Hr_sig_b)
Hr_sig_b = BatchNormalization()(Hr_sig_b) 
Hr_sig_b = MaxPooling1D(3, 3)(Hr_sig_b)
```

```python
Hr_sig2 = Input(shape=(600,6))

Hr_sig_c = Conv1D(16, 6, activation='relu')(Hr_sig2)
Hr_sig_c = BatchNormalization()(Hr_sig_c) 
Hr_sig_c = MaxPooling1D(3, 3)(Hr_sig_c) 

Hr_sig_c = Conv1D(32, 6, activation='relu')(Hr_sig_c)
Hr_sig_c = BatchNormalization()(Hr_sig_c) 
Hr_sig_c = MaxPooling1D(3, 3)(Hr_sig_c) 

Hr_sig_c = Conv1D(64, 6, activation='relu')(Hr_sig_c)
Hr_sig_c = BatchNormalization()(Hr_sig_c) 
Hr_sig_c = MaxPooling1D(3, 3)(Hr_sig_c)

Hr_sig_c = Conv1D(128, 6, activation='relu')(Hr_sig_c)
Hr_sig_c = BatchNormalization()(Hr_sig_c) 
Hr_sig_c = MaxPooling1D(3, 3)(Hr_sig_c)
```

```python
Hr_sig3 = Input(shape=(600,6))

Hr_sig_d = Conv1D(16, 6, activation='relu')(Hr_sig3)
Hr_sig_d = BatchNormalization()(Hr_sig_d) 
Hr_sig_d = MaxPooling1D(3, 3)(Hr_sig_d) 

Hr_sig_d = Conv1D(32, 6, activation='relu')(Hr_sig_d)
Hr_sig_d = BatchNormalization()(Hr_sig_d) 
Hr_sig_d = MaxPooling1D(3, 3)(Hr_sig_d) 

Hr_sig_d = Conv1D(64, 6, activation='relu')(Hr_sig_d)
Hr_sig_d = BatchNormalization()(Hr_sig_d) 
Hr_sig_d = MaxPooling1D(3, 3)(Hr_sig_d)

Hr_sig_d = Conv1D(128, 6, activation='relu')(Hr_sig_d)
Hr_sig_d = BatchNormalization()(Hr_sig_d) 
Hr_sig_d = MaxPooling1D(3, 3)(Hr_sig_d)
```

```python
Hr_sig4 = Input(shape=(600,6), name='sig4_input_new')

Hr_sig_e = Conv1D(16, 6, activation='relu')(Hr_sig4)
Hr_sig_e = BatchNormalization()(Hr_sig_e) 
Hr_sig_e = MaxPooling1D(3, 3)(Hr_sig_e) 

Hr_sig_e = Conv1D(32, 6, activation='relu')(Hr_sig_e)
Hr_sig_e = BatchNormalization()(Hr_sig_e) 
Hr_sig_e = MaxPooling1D(3, 3)(Hr_sig_e) 

Hr_sig_e = Conv1D(64, 6, activation='relu')(Hr_sig_e)
Hr_sig_e = BatchNormalization()(Hr_sig_e) 
Hr_sig_e = MaxPooling1D(3, 3)(Hr_sig_e)

Hr_sig_e = Conv1D(128, 6, activation='relu')(Hr_sig_e)
Hr_sig_e = BatchNormalization()(Hr_sig_e) 
Hr_sig_e = MaxPooling1D(3, 3)(Hr_sig_e)
```

```python
x = keras.layers.Concatenate()([Hr_sig_a, Hr_sig_b, Hr_sig_c, Hr_sig_d, Hr_sig_e])
x = Flatten()(x)
x = layers.Dense(128, activation='relu')(x)
x = layers.Dense(64, activation='relu')(x)
x = layers.Dropout(0.5)(x)
x = layers.Dense(2, activation='sigmoid')(x)

x.shape
```

```python
model = Model(inputs=[Hr_sig, Hr_sig1, Hr_sig2, Hr_sig3, Hr_sig4], outputs=x)
```
